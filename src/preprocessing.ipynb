# LIMPIEZA DE DATOS INICIAL Y VISUALIZACION "maestro_clientes.csv"

```python
import pandas as pd
import numpy as np

df_maestro_clientes = pd.read_csv("../Datos/Raw/maestro_clientes.csv", sep=";")
print(f"Filas estrictamente duplicadas: {df_maestro_clientes.duplicated().sum()}")
df_maestro_clientes.drop_duplicates(inplace=True)
print(f"Las columnas del documento maestro_clientes.csv son: {df_maestro_clientes.columns.tolist()}")
print(f"Filas totales: {len(df_maestro_clientes)}")
print("---------------------------------------------------------")
print(f"Ejemplo de la base de datos:")
print(f"\n {df_maestro_clientes.sample(5)}")
print("---------------------------------------------------------")
print(f"Tipos de cliente: {df_maestro_clientes['Tipo Cliente'].unique()}")
```

```python
df_no_empresa = df_maestro_clientes[df_maestro_clientes['Tipo Cliente'] != 'Empresa']
datos_na_previos = df_no_empresa[df_no_empresa['Edad'].isna()]

if not datos_na_previos.empty:
    print("Filas de Consumidor Final con 'Edad' faltante antes de limpiar:")
    print(datos_na_previos)
else:
    print("No hay Consumidores Finales con 'Edad' faltante.")

# Calcular la mediana de edad de los Consumidores Finales (ignorando NaN)
mediana_consumidor = df_maestro_clientes[df_maestro_clientes['Tipo Cliente'] == 'Consumidor final']['Edad'].median()

# Asignar -1 a las Empresas que tienen Edad NaN
df_maestro_clientes.loc[(df_maestro_clientes['Tipo Cliente'] == 'Empresa') & (df_maestro_clientes['Edad'].isna()),'Edad'] = -1

# Asignar la mediana a los Consumidores Finales que tienen Edad NaN
df_maestro_clientes.loc[(df_maestro_clientes['Tipo Cliente'] == 'Consumidor final') & (df_maestro_clientes['Edad'].isna()), 'Edad'] = mediana_consumidor

print("\nFilas de Consumidor Final con 'Edad' faltante después de limpiar:")
print(df_maestro_clientes.loc[df_maestro_clientes['Cliente'] == 'cliente_468723'])
```

# VISUALIZACION DEL ESTADO Y CONGRUENCIA DE LAS BASES DE DATOS INICIALES

```python
import numpy as np
# -----------------------------------------------------------------------------
# 1. INTEGRIDAD ESTRUCTURAL Y CONCURRENCIA DE DATOS
# -----------------------------------------------------------------------------
print("--- 1. INTEGRIDAD ESTRUCTURAL Y REFERENCIAL ---")

# Singularidad de Cliente en Maestro
total_filas_m = len(df_maestro_clientes)
clientes_unicos_m = df_maestro_clientes["Cliente"].nunique()
clientes_duplicados = df_maestro_clientes[df_maestro_clientes["Cliente"].duplicated(keep=False)]["Cliente"].unique()
print(f"Maestro Clientes: {total_filas_m} filas | {clientes_unicos_m} clientes únicos.")
if total_filas_m != clientes_unicos_m:
    print(f"Hay {len(clientes_duplicados)} IDs de clientes duplicados en el maestro.")

# Integridad Referencial
set_clientes_m = set(df_maestro_clientes["Cliente"].dropna())
set_clientes_t = set(df_base_transaccional["Cliente"].dropna())

clientes_solo_tx = set_clientes_t - set_clientes_m
clientes_sin_compras = set_clientes_m - set_clientes_t

print(f"Clientes en Transaccional pero NO en Maestro: {len(clientes_solo_tx)} ({len(clientes_solo_tx)/len(set_clientes_t):.2%})")
print(f"Clientes en Maestro pero SIN Transacciones: {len(clientes_sin_compras)} ({len(clientes_sin_compras)/len(set_clientes_m):.2%})")

# Tratamiento Datos
print("\nDistribución de 'Tratamiento Datos' en Maestro:")
print(df_maestro_clientes["Tratamiento Datos"].value_counts(dropna=False, normalize=True)* 100)

# -----------------------------------------------------------------------------
# 2. ANÁLISIS DE COMPLETITUD Y VALORES RELEVANTES/FALTANTES
# -----------------------------------------------------------------------------
print("\n--- 2. COMPLETITUD DE VARIABLES (NULOS Y VALORES ESPECIALES) ---")

def reporte_nulos(df, nombre_df):
    reporte = pd.DataFrame({
"Total Nulos": df.isna().sum(),"% Nulos": (df.isna().sum() / len(df)) * 100, "Valores Únicos": df.nunique()})
    # Detectar cadenas especiales como '(No Value)'
    no_value_counts = {}
    for col in df.select_dtypes(include="object").columns:
        cnt = (df[col] == "(No Value)").sum()
        if cnt > 0:
            no_value_counts[col] = cnt
    reporte["Cantidad '(No Value)'"] = pd.Series(no_value_counts)
    reporte["Cantidad '(No Value)'"] = reporte["Cantidad '(No Value)'"].fillna(0)
    print(f"\n[Reporte de Calidad: {nombre_df}]")
    print(reporte[reporte["Total Nulos"] > 0 | (reporte["Cantidad '(No Value)'"] > 0)])


reporte_nulos(df_maestro_clientes, "Maestro Clientes")
reporte_nulos(df_base_transaccional, "Base Transaccional")

# -----------------------------------------------------------------------------
# 3. PERFILAMIENTO DEMOGRÁFICO Y CONSISTENCIA (MAESTRO CLIENTES)
# -----------------------------------------------------------------------------
print("\n--- 3. PERFILAMIENTO DEMOGRÁFICO DE CLIENTES ---")

print("Distribución de Tipo Cliente:")
print(df_maestro_clientes["Tipo Cliente"].value_counts(dropna=False))

print("\nResumen Estadístico de 'Edad' por 'Tipo Cliente':")
print(df_maestro_clientes.groupby("Tipo Cliente")["Edad"].describe(percentiles=[0.01, 0.05, 0.5, 0.95, 0.99]))
print("\nDistribución de 'Genero' por 'Tipo Cliente':")
print(pd.crosstab(df_maestro_clientes["Tipo Cliente"], df_maestro_clientes["Genero"], dropna=False, margins=True))

# Verificación de fechas
df_maestro_clientes["Primera Fecha"] = pd.to_datetime(df_maestro_clientes["Primera Fecha"], errors="coerce")
print(f"\nRango de 'Primera Fecha' en Maestro: Desde {df_maestro_clientes['Primera Fecha'].min()} Hasta {df_maestro_clientes['Primera Fecha'].max()}")

# -----------------------------------------------------------------------------
# 4. PERFILAMIENTO TRANSACCIONAL Y REGLAS DE NEGOCIO
# -----------------------------------------------------------------------------
print("\n--- 4. CONSISTENCIA ARITMÉTICA Y REGISTROS TRANSACCIONALES ---")

df_base_transaccional["Fecha Pedido"] = pd.to_datetime(df_base_transaccional["Fecha Pedido"], errors="coerce")
print(f"Rango de 'Fecha Pedido': Desde {df_base_transaccional['Fecha Pedido'].min()} Hasta {df_base_transaccional['Fecha Pedido'].max()}")

# Validación de la regla: Valor = Cantidad * Precio
# Creamos un cálculo teórico
calculado = df_base_transaccional["Cantidad"] * df_base_transaccional["Precio"]
diferencia = np.abs(df_base_transaccional["Valor"] - calculado)

# Consideramos discrepancia si la diferencia es mayor a 1 peso (tolerancia de redondeo)
discrepancias = df_base_transaccional[diferencia > 1.0]
print(f"Transacciones con discrepancia aritmética (|Valor - Cantidad*Precio| > 1): {len(discrepancias)} ({len(discrepancias)/len(df_base_transaccional):.2%})")

# Registros con valores Cero o Negativos
print("\nRegistros con valores <= 0:")
print(f"  Cantidad <= 0: {(df_base_transaccional['Cantidad'] <= 0).sum()}")
print(f"  Precio <= 0:   {(df_base_transaccional['Precio'] <= 0).sum()}")
print(f"  Valor <= 0:    {(df_base_transaccional['Valor'] <= 0).sum()}")

# -----------------------------------------------------------------------------
# 5. ESTRUCTURA DE CATALOGO, GEOGRAFÍA Y CANALES
# -----------------------------------------------------------------------------
print("\n--- 5. CARDINALIDAD Y ESTRUCTURA DE NEGOCIO ---")

print("Cardinalidad de Categorías de Producto:")
print(f"  Negocios: {df_base_transaccional['Negocio'].nunique()}")
print(f"  Categorías: {df_base_transaccional['Categoria'].nunique()}")
print(f"  Subcategorías: {df_base_transaccional['Subategoria'].nunique()}")
print(f"  Productos Únicos: {df_base_transaccional['Producto'].nunique()}")

print("\nConcentración Transaccional por Negocio:")
print(df_base_transaccional["Negocio"].value_counts(normalize=True) * 100)

print("\nTop 5 Regiones por número de transacciones:")
print(df_base_transaccional["Region"].value_counts().head(5))

# -------------------------------------------------------------------------
# 6. CRUCE TEMPORAL: CONSISTENCIA PRIMERA FECHA vs PRIMER PEDIDO
# -------------------------------------------------------------------------
primer_pedido_tx = (df_base_transaccional.groupby("Cliente")["Fecha Pedido"].min())

# Tomamos una sola Primera Fecha por cliente
primera_fecha_maestro = (df_maestro_clientes.drop_duplicates("Cliente").set_index("Cliente")["Primera Fecha"])

# Cruce de fechas
comparacion = pd.DataFrame({"Primera Fecha": primera_fecha_maestro,"Primer Pedido Real": primer_pedido_tx})

# Solo clientes duplicados
comparacion = comparacion.loc[comparacion.index.isin(clientes_duplicados)]

# Condición que busca clientes cuya 'Primera Fecha' en Maestro es posterior a su 'Primer Pedido Real' en Transaccional
resultado = comparacion[comparacion["Primera Fecha"] > comparacion["Primer Pedido Real"]]

print(f"Clientes que están duplicados en Maestro Y cuya 'Primera Fecha' es posterior a su 'Primer Pedido Real': {len(resultado)}")
```

# Feature engineering inicial: Creación de las primeras variables explicativas y limpieza final de datos


```python
# Convertir 'Primera Fecha' a datetime
df_maestro_clientes["Primera Fecha"] = pd.to_datetime(df_maestro_clientes["Primera Fecha"], errors="coerce")

# Ordenar para conservar la primera fecha real y el perfil más reciente
df_maestro_clientes = df_maestro_clientes.sort_values(by=["Cliente", "Primera Fecha"], ascending=[True, True])

# Agrupar por ID de cliente para eliminar duplicados de ID
df_maestro_consolidado = df_maestro_clientes.groupby("Cliente", as_index=False).agg(
    {"Tipo Cliente": "last", "Edad": "last","Genero": "last", "Tratamiento Datos": "last", "Primera Fecha": "first"})

# Crear variable de Grupo Generacional
def categorizar_generacion(row):
    if row["Tipo Cliente"] == "Empresa" or row["Edad"] == -1:
        return "Empresa"
    edad = row["Edad"]
    if edad < 29: #1997-2012
        return "Gen Z"
    elif edad <= 42: #1981-1996
        return "Millennials"
    elif edad <= 58: #1965-1980
        return "Gen X"
    else:
        return "Boomers"

df_maestro_consolidado["Grupo_Generacional"] = df_maestro_consolidado.apply(categorizar_generacion, axis=1)

# Crear bandera de Habeas Data
df_maestro_consolidado["Permite_Contacto_HabeasData"] = (
    df_maestro_consolidado["Tratamiento Datos"].astype(str).str.strip().str.upper().apply(lambda x: 1 if x == "SÍ" else 0)
)

print(f"Maestro consolidado por ID unico: {len(df_maestro_consolidado)} clientes.")
print("\nDistribución por Grupo Generacional:")
print(df_maestro_consolidado["Grupo_Generacional"].value_counts())
```

```python
# Convertir 'Fecha Pedido' a datetime
df_base_transaccional["Fecha Pedido"] = pd.to_datetime(df_base_transaccional["Fecha Pedido"], errors="coerce")

# Remplazar '(No Value)' en región
df_base_transaccional["Region"] = df_base_transaccional["Region"].replace("(No Value)", "DESCONOCIDO")

# Identificar devoluciones o ajustes (Valor <= 0 o Cantidad <= 0)
df_base_transaccional["Es_Devolucion"] = (df_base_transaccional["Valor"] <= 0) | (df_base_transaccional["Cantidad"] <= 0)

print(f"Transacciones procesadas: {len(df_base_transaccional)} filas totales.")
print(f"Movimientos que son devoluciones/ajustes: {df_base_transaccional['Es_Devolucion'].sum()}.")
```

```python
# Cruce de la base transaccional con el maestro consolidado
df_transacciones_completa = pd.merge(df_base_transaccional,
    df_maestro_consolidado[["Cliente","Tipo Cliente","Edad","Genero","Grupo_Generacional","Permite_Contacto_HabeasData","Primera Fecha",]],
    on="Cliente", how="left",)

print(f"Base unificada final: {len(df_transacciones_completa)} filas.")
print(
    f"Ejemplo de la base unificada:\n{df_transacciones_completa.head(2)}"
)
```

# Creación del archivo preprocesado y unificado

```python
import json
from pathlib import Path

RUTA_DATOS = Path("../Datos/Pre-processed")
RUTA_DATOS.mkdir(parents=True, exist_ok=True)

NOMBRE_BASE = "base_unificada_transaccional_clientes"
ruta_csv = RUTA_DATOS / f"{NOMBRE_BASE}.csv"
ruta_parquet = RUTA_DATOS / f"{NOMBRE_BASE}.parquet"
ruta_meta = RUTA_DATOS / f".{NOMBRE_BASE}.meta.json"

def exportar_base_unificada(df):
    meta_actual = {"filas": len(df), "columnas": list(df.columns)}

    # Verificar si ya existen los archivos y coinciden
    if ruta_csv.exists() and ruta_parquet.exists() and ruta_meta.exists():
        try:
            meta_previa = json.loads(ruta_meta.read_text(encoding="utf-8"))
            if meta_previa == meta_actual:
                print("Sin cambios: La base unificada ya existe con la misma estructura y filas.")
                return
        except (json.JSONDecodeError, OSError):
            pass

    print(f"Exportando base unificada ({len(df):,} filas)")
    df.to_parquet(ruta_parquet, index=False)
    df.to_csv(ruta_csv, sep=";", index=False, encoding="utf-8-sig", date_format="%Y-%m-%d")

    ruta_meta.write_text(json.dumps(meta_actual, ensure_ascii=False), encoding="utf-8")
    print(f"Archivos actualizados en {RUTA_DATOS.resolve()}")

exportar_base_unificada(df_transacciones_completa)
```
